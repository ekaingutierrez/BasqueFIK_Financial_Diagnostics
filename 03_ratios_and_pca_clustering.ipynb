# ratios

```python
df["FM"]= df["Activo circulante\nmil EUR\nÚlt. año disp."] - df["Pasivo líquido\nmil EUR\nÚlt. año disp."]
df["Activo Corriente Operativo"]= df["Existencias\nmil EUR\nÚlt. año disp."] + df["Deudores\nmil EUR\nÚlt. año disp."]
df["Pasivo Corriente Operativo"]= df["Acreedores comerciales\nmil EUR\nÚlt. año disp."]
df["NOF"]=df["Activo Corriente Operativo"]-df["Pasivo Corriente Operativo"]

#liquidez

df["RATIO DE TESORERIA INMEDIATA"] = df["Tesorería\nmil EUR\nÚlt. año disp."] / df["Pasivo líquido\nmil EUR\nÚlt. año disp."]


df["RATIO DE PRUEBA ACIDA"] = (df["Activo circulante\nmil EUR\nÚlt. año disp."] - df["Existencias\nmil EUR\nÚlt. año disp."]) / df["Pasivo líquido\nmil EUR\nÚlt. año disp."]


df["RATIO DE LIQUIDEZ"] = df["Activo circulante\nmil EUR\nÚlt. año disp."] / df["Pasivo líquido\nmil EUR\nÚlt. año disp."]

#solvencia

df["RATIO DE SOLVENCIA"] = df["Total activo\nmil EUR\nÚlt. año disp."] / (df["Pasivo fijo\nmil EUR\nÚlt. año disp."] + df["Pasivo líquido\nmil EUR\nÚlt. año disp."])


df["RATIO DE ENDEUDAMIENTO"] = (df["Pasivo fijo\nmil EUR\nÚlt. año disp."] + df["Pasivo líquido\nmil EUR\nÚlt. año disp."]) / df["Total pasivo y capital propio\nmil EUR\nÚlt. año disp."] 


df["RATIO DE INDEPENDENCIA"] = df["Total pasivo y capital propio\nmil EUR\nÚlt. año disp."] - (df["Pasivo fijo\nmil EUR\nÚlt. año disp."] + df["Pasivo líquido\nmil EUR\nÚlt. año disp."]) / df["Total pasivo y capital propio\nmil EUR\nÚlt. año disp."]  


df["RATIO DE APALANCAMIENTO FINANCIERO (%)"] = (df["Deudas financieras\nmil EUR\nÚlt. año disp."] / df["Fondos propios\nmil EUR\nÚlt. año disp."])*100

#tangibilidad e intangibilidad

df["RATIO DE TANGIBILIDAD"] = df["Inmovilizado material\nmil EUR\nÚlt. año disp."] / df["Total activo\nmil EUR\nÚlt. año disp."]


df["RATIO DE INTANGIBILIDAD"] = df["Inmovilizado inmaterial\nmil EUR\nÚlt. año disp."] / df["Total activo\nmil EUR\nÚlt. año disp."]

#económicos 

df["RATIO DE ENDEUDAMIENTO DEL EBITDA"] = df["Deudas financieras\nmil EUR\nÚlt. año disp."] / df["EBITDA\nmil EUR\nÚlt. año disp."]


df["RATIO DE ENDEUDAMIENTO SOBRE EL EBIT"] = df["Deudas financieras\nmil EUR\nÚlt. año disp."] / df["Resultado Explotación\nmil EUR\nÚlt. año disp."]


df["COSTE MEDIO DE LA DEUDA"] = df["Gastos financieros\nmil EUR\nÚlt. año disp."] / df["Deudas financieras\nmil EUR\nÚlt. año disp."]


df["DSRC (Deuda Servicio Ratio de Cobertura)"] = df["EBITDA\nmil EUR\nÚlt. año disp."] / df["Gastos financieros\nmil EUR\nÚlt. año disp."]

# rentabilidad
df["Margen bruto sobre ventas"]=(df["Importe neto Cifra de Ventas\nmil EUR\nÚlt. año disp."]+ df["Existencias\nmil EUR\nÚlt. año disp."]-df["Consumo de mercaderías y de materias\nmil EUR\nÚlt. año disp."])/df["Importe neto Cifra de Ventas\nmil EUR\nÚlt. año disp."]

df["Margen sobre EBITDA"]=df["EBITDA\nmil EUR\nÚlt. año disp."]/df["Importe neto Cifra de Ventas\nmil EUR\nÚlt. año disp."]

df["ROA"] = (df["Resultado Explotación\nmil EUR\nÚlt. año disp."] / df["Total activo\nmil EUR\nÚlt. año disp."])*100

df["ROE"] =  df["Resultado del Ejercicio\nmil EUR\nÚlt. año disp."] / df["Fondos propios\nmil EUR\nÚlt. año disp."] * 100

#eficiencia operativa
dia_de_cobro= df["Deudores\nmil EUR\nÚlt. año disp."] / df["Importe neto Cifra de Ventas\nmil EUR\nÚlt. año disp."] * 365
dia_de_pago = df["Acreedores comerciales\nmil EUR\nÚlt. año disp."] / df["Consumo de mercaderías y de materias\nmil EUR\nÚlt. año disp."] * 365

        # el abs funciona para que no haya valores negativos en la columna
df["CONVERSION DE EFECTIVO"] = abs(dia_de_cobro - dia_de_pago)

```

# PCA (KMEANS)

separar X

```python
X= df[["Número empleados\nÚlt. año disp.", 'RATIO DE TESORERIA INMEDIATA',
       'RATIO DE PRUEBA ACIDA', 'RATIO DE LIQUIDEZ', 'RATIO DE SOLVENCIA',
       'RATIO DE ENDEUDAMIENTO', 'RATIO DE INDEPENDENCIA', 'RATIO DE TANGIBILIDAD',
       'RATIO DE INTANGIBILIDAD', 'RATIO DE ENDEUDAMIENTO DEL EBITDA', 'COSTE MEDIO DE LA DEUDA',
       'DSRC (Deuda Servicio Ratio de Cobertura)', 'ROA', 'ROE']]
y=df["Nombre"]
```

estandarizar

```python
# Usamos RobustScaler para mitigar los outliers desde el principio, antes del PCA
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
data_scaled = scaler.fit_transform(X)
```

```python
pca=PCA(n_components=2)
principal_components= pca.fit_transform(data_scaled)
```

```python
print(f"Variación explicada por componente: {pca.explained_variance_ratio_}")
print(f"Coeficientes del primer componente PC1: {pca.components_[0]}")
print(f"Coeficientes del segundo componente PC2: {pca.components_[1]}")

#en total los dos componentes explican un 30% de los datos, osea que no explican suficientemente bien los datos
```

pca (numero de componentes óptimos)

```python
#numero de componentes optimos
pca=PCA()
principal_components= pca.fit_transform(data_scaled)

cumulative_variance_ratio=np.cumsum(pca.explained_variance_ratio_) #se suman en cumsum porque nos interesa saber cuando parar
plt.plot(range(1,len(cumulative_variance_ratio) + 1), cumulative_variance_ratio, marker="o")
plt.xlabel("Numero de Componentes Principales")
plt.ylabel("Varianza Explicada Acumulativa")
plt.title("Varianza Explicada Acumulativa vs. Numero de Componentes Principales")
plt.show()
```

```python
n_components=np.argmax(cumulative_variance_ratio >= 0.95) + 1
print(f"Numero óptimo de componentes principales: {n_components}")
```

```python
pca_final = PCA(n_components=n_components) 
principal_components = pca_final.fit_transform(data_scaled)

df_pca = pd.DataFrame(data=principal_components, columns=[f"PC{i+1}" for i in range(n_components)])
display(df_pca.head())

# Asignamos directamente la matriz limpia sin manipular matemáticamente las componentes
X_pca_limpio = principal_components
```

clusters para kmeans

```python
dist = []
K = range(1,15) 
for k in K:
    kmeanModel = KMeans(n_clusters =k, random_state=42)
    kmeanModel.fit(X_pca_limpio)
    dist.append(kmeanModel.inertia_)
```

```python
plt.figure(figsize=(10,6))
plt.plot(K,dist, marker="o")
plt.xlabel("Numero de clusters")
plt.ylabel("Inertia")
plt.title("Metodo del codo")
plt.grid(True)
plt.show()
```

```python
# kmeans óptimo
optimal_k=5
kmeans=KMeans(n_clusters=optimal_k, random_state=42)
kmeans.fit(X_pca_limpio)        #entreno con el que no tiene outliers y predigo con el que si
y_pred=kmeans.predict(df_pca)
```

```python
y_pred
```

```python
# ver cuantos hay de cada cluster
df["Cluster"]=y_pred
df.head()
```

```python
df["Cluster"].value_counts()
```

```python
# metrica silhouette para ver como de bien separados estan los clusters y si dentro los puntos están compactos
score= silhouette_score(df_pca,y_pred)
print(f"Silhouette score:{score:3f}")
#no es una mala puntuacion para los datos que tenemos
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X_pca_limpio[y_pred==0, 0], X_pca_limpio[y_pred==0, 1], s=100, c='red', label ='Cluster 0')
plt.scatter(X_pca_limpio[y_pred==1, 0], X_pca_limpio[y_pred==1, 1], s=100, c='blue', label ='Cluster 1')
plt.scatter(X_pca_limpio[y_pred==2, 0], X_pca_limpio[y_pred==2, 1], s=100, c='green', label ='Cluster 2')
plt.scatter(X_pca_limpio[y_pred==3, 0], X_pca_limpio[y_pred==3, 1], s=100, c='yellow', label ='Cluster 3')
plt.scatter(X_pca_limpio[y_pred==4, 0], X_pca_limpio[y_pred==4, 1], s=100, c='pink', label ='Cluster 4')
plt.legend()
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Clusters sobre PCA')
plt.show() #los datos se distribuyen mas horizontalmente porque pc1 explica el mas varianza que pc2
```
